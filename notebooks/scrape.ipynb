```python
from curl_cffi import requests
from bs4 import BeautifulSoup
import time

# Create a session that impersonates a real Chrome browser TLS fingerprint
session = requests.Session(impersonate="chrome120")

def getAndParseURL(url):
    headers = {
        "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8",
        "Accept-Language": "en-US,en;q=0.9",
        "Referer": "https://sofifa.com/",
        "Sec-Ch-Ua": '"Not_A Brand";v="8", "Chromium";v="120", "Google Chrome";v="120"',
        "Sec-Ch-Ua-Mobile": "?0",
        "Sec-Ch-Ua-Platform": '"Windows"',
        "Sec-Fetch-Dest": "document",
        "Sec-Fetch-Mode": "navigate",
        "Sec-Fetch-Site": "same-origin",
        "Sec-Fetch-User": "?1",
        "Upgrade-Insecure-Requests": "1"
    }
    
    try:
        response = session.get(url, headers=headers, timeout=10)
        
        if response.status_code != 200:
            print(f"HTTP Error {response.status_code} for URL: {url}")
            return None
            
        soup = BeautifulSoup(response.text, "html.parser")
        return soup
        
    except Exception as e:
        print(f"Request failed: {e}")
        return None


# Scraping Execution
players = []

for page in pages:
    print(f"Scraping: {page}")
    html = getAndParseURL(page)
    
    if html is None:
        continue

    # Locate the table container
    table = html.find("table") or html.select_one(".table")
    
    if table is not None:
        # Extract table headers once per page
        header_tr = table.find("tr")
        headers = [th.get_text().strip() for th in header_tr.find_all("th")] if header_tr else []

        # Parse player rows
        rows = table.find_all("tr")[1:]
        for row in rows:
            cols = row.find_all("td")
            if len(cols) > 1:
                player = {}
                player["name"] = cols[1].get_text().strip()
                
                for i in range(2, len(cols)):
                    if i < len(headers) and headers[i]:
                        player[headers[i]] = cols[i].get_text().strip()
                
                players.append(player)
    else:
        print(f"Table not found on page: {page}")
    
    time.sleep(2)  # Delay between requests to prevent rate-limiting

print(f"Finished! Scraped {len(players)} total players.")
```
